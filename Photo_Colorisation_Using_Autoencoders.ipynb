```python
# Run this cell first!
import torch
print("GPU available:", torch.cuda.is_available())

from google.colab import drive
drive.mount('/content/drive')

# Set your directory structure - adjust if your Drive structure is different!
base_dir = ""

# Train folders (images 1-100)
train_gray_dir  = f"{base_dir}/train/input"   # img1_grayscale.jpg ... img100_grayscale.jpg
train_color_dir = f"{base_dir}/train/output"  # img1_color.jpg ... img100_color.jpg

# Test folders (images 101-150)
test_gray_dir  = f"{base_dir}/test/input"     # img101_grayscale.jpg ... img150_grayscale.jpg
test_color_dir = f"{base_dir}/test/output"    # img101_color.jpg ... img150_color.jpg

print("Train grayscale folder:", train_gray_dir)
print("Train color folder    :", train_color_dir)
print("Test grayscale folder :", test_gray_dir)
print("Test color folder     :", test_color_dir)


import cv2
import numpy as np
import os

img_size = 120

def load_images(folder, to_gray=False, to_rgb=False):
    files = sorted([f for f in os.listdir(folder) if f.lower().endswith(('.jpg', '.png', '.jpeg'))])
    images = []
    for fname in files:
        path = os.path.join(folder, fname)
        img = cv2.imread(path)
        if img is not None:
            if to_gray:
                img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            elif to_rgb:
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img, (img_size, img_size))
            if to_gray:
                img = np.expand_dims(img, axis=-1)
            images.append(img.astype('float32') / 255.0)
    return np.array(images), files

# Load training and test data
train_gray, train_gray_files = load_images(train_gray_dir, to_gray=True)
train_color, train_color_files = load_images(train_color_dir, to_rgb=True)

test_gray, test_gray_files = load_images(test_gray_dir, to_gray=True)
test_color, test_color_files = load_images(test_color_dir, to_rgb=True)

print("Train grayscale:", train_gray.shape)
print("Train color   :", train_color.shape)
print("Test grayscale:", test_gray.shape)
print("Test color   :", test_color.shape)


# Make sure shapes match
print("Train grayscale files:", train_gray_files[:3], "...")
print("Train color files    :", train_color_files[:3], "...")

# Inputs (X), Outputs (Y)
X_train = train_gray   # shape: (100, 120, 120, 1)
Y_train = train_color  # shape: (100, 120, 120, 3)

X_test = test_gray     # shape: (50, 120, 120, 1)
Y_test = test_color    # shape: (50, 120, 120, 3)

print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
print("X_test shape :", X_test.shape)
print("Y_test shape :", Y_test.shape)


from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, Conv2DTranspose
from tensorflow.keras.optimizers import Adam

input_layer = Input(shape=(img_size, img_size, 1))

# Encoder
x = Conv2D(64, (3,3), activation='relu', padding='same')(input_layer)
x = Conv2D(128, (3,3), activation='relu', padding='same', strides=2)(x)
x = Conv2D(256, (3,3), activation='relu', padding='same', strides=2)(x)

# Decoder
x = Conv2DTranspose(128, (3,3), activation='relu', padding='same', strides=2)(x)
x = Conv2DTranspose(64, (3,3), activation='relu', padding='same', strides=2)(x)
output_layer = Conv2D(3, (3,3), activation='sigmoid', padding='same')(x)

autoencoder = Model(inputs=input_layer, outputs=output_layer)
autoencoder.compile(optimizer=Adam(1e-3), loss='mse')
autoencoder.summary()


history = autoencoder.fit(
    X_train, Y_train,
    validation_split=0.1,
    epochs=500,
    batch_size=16,
    verbose=1
)


autoencoder.save(f"{base_dir}/colorization_autoencoder.keras")
print("Model saved to Drive.")



from tensorflow.keras.models import load_model
model_path = f"{base_dir}/colorization_autoencoder.keras"
autoencoder = load_model(model_path)
print("Model loaded from:", model_path)



import matplotlib.pyplot as plt

# Change the index below to test any image from test folder (0 to 49)
test_idx = 0
test_img_gray = X_test[test_idx]

# Model expects shape (1, 120, 120, 1)
test_img_input = np.expand_dims(test_img_gray, axis=0)
pred = autoencoder.predict(test_img_input)
pred = np.clip(pred[0], 0, 1)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(test_img_gray.squeeze(), cmap='gray')
plt.title("Grayscale Input (Test)")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(pred)
plt.title("Colorized Output")
plt.axis('off')

plt.show()


```
